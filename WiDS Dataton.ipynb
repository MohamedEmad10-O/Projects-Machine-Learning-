```python
from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder(sparse_output=False)
building_class_encoded = ohe.fit_transform(df[["building_class"]])

encoded_df = pd.DataFrame(building_class_encoded, columns=ohe.get_feature_names_out(["building_class"]), index=df.index)

df = df.drop(columns=["building_class"]).join(encoded_df)
```

```python
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df["State_Factor"] = le.fit_transform(df["State_Factor"])
df["facility_type"] = le.fit_transform(df["facility_type"])
```

```python
df.head()
```

```python
X = df.drop(columns= ["State_Factor"] , axis = 1)
y = df["State_Factor"]
```

```python
from sklearn.model_selection import train_test_split

X_train , X_test , y_train , y_test = train_test_split(X , y , test_size = 0.3 , random_state = 42)
```

```python
from sklearn.preprocessing import StandardScaler
scalar = StandardScaler()
X[["floor_area"]] = scalar.fit_transform(X[["floor_area"]])
```

```python
for col in X_train.select_dtypes([int , float]).columns :
  X_train[[col]] = scalar.fit_transform(X_train[[col]])
```

```python
for col in X_test.select_dtypes([int , float]).columns :
  X_test[[col]] = scalar.fit_transform(X_test[[col]])
```

```python
df.head()
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
model.fit(X_train , y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score , confusion_matrix , classification_report, roc_curve, auc , roc_auc_score
```

```python
report = classification_report(y_test, y_pred)
```
